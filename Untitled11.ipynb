```python
import pandas as pd

file_path = '/content/drive/MyDrive/kanoune-kessou-fin2-tp2/extportal.xls'

try:
    # Charger le fichier en ignorant les 4 premières lignes de métadonnées et en spécifiant le séparateur tabulation (\t)
    df = pd.read_csv(file_path, sep='\t', skiprows=4)
    print("La base de données a été importée et nettoyée avec succès !")
    # Affichage du tableau formaté
    display(df)
except Exception as e:
    print(f"Erreur lors du chargement propre de la base de données : {e}")
```

```python
# Étape de nettoyage préalable : Conversion des colonnes de texte avec virgules en valeurs numériques
cols_to_convert = ['Securities', 'Nbr 1 Account', 'Nbr 2 Account', 'CVA Account', 'CVA Nbr 2 Account', 'Total Holdings']

for col in cols_to_convert:
    if col in df.columns:
        # On supprime les virgules et espaces puis convertit en numérique
        df[col] = pd.to_numeric(df[col].astype(str).str.replace(',', ''), errors='coerce')

print("--- Requête de calcul 1 : Somme totale cumulée par type de compte ---")
sommes = df[cols_to_convert[:-1]].sum()
display(sommes.to_frame(name='Somme Totale (SDRs)'))

print("\n--- Requête de calcul 2 : Moyenne globale des avoirs totaux (Total Holdings) ---")
moyenne_holdings = df['Total Holdings'].mean()
print(f"La moyenne des Total Holdings est de : {moyenne_holdings:,.2f} SDRs\n")

print("--- Requête de calcul 3 : Top 5 des dates avec les avoirs totaux les plus élevés ---")
top_5_dates = df.nlargest(5, 'Total Holdings')[['Date', 'Total Holdings']]
display(top_5_dates)
```
